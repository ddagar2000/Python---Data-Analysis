# Object Oriented Programming 

oops is a programming paradigm based on the concept of "objects", which can contain data in the form of fields (often known as attributes/ properties) and code in the form of procedures (often known as methods)

Key concepts of oops:
1. Class: A blueprint for creating objects. A class encapsulates data for the object and methods to manipulate that data.
2. Object: An instance of a class. It is a self - contained unit that consists of both data and procedures to manipulate the data.
3. Attributes: Data stored inside an object. Attributes are defined in the class.
4. Methods: Functions defined inside a class that describe the behaviors of the objects.

Procedural Programming vs Objects Oriented Programming:

Procedural Programming
Approach: Focuses on functions or procedures to operate on data.
Structure: Code is organized into functions, and data is passed between these functions.
Reusability: Code reusability can be achieved through function reuse, but managing large codebases can become complex.
State Management: functions do not retain state: state is managed through variables passed between fuctions.
Example Languages: C, pascal


Approach: Focuses on objects conatin both data and methods.
Structure: Code is organized into classes and objects, promoting encapsulation.
Reusability: Promotes code reuse through inheritance and polymorphism, making it easier to manage larger codebases.
State Management: Objects retain state, making it easier to model real-world entities.
Example Languages: Java, Python, C++

OOP Principles:

1. Encapsulation:

Definition: Building data and methods that operate on the data into a single unit, or class.
Benefits: Encapsulation hides the internal state of the object from the outside world and only exposes a controlled interface. Thsi leads to better modularity and protects the integrity of the objects data.

2. Abstraction:

Definition: Hiding the conplex implementation details and showing only the necessary features of the objects.
Benefits: Simplifies the interface to interact with the object making it easier to use and understand without needing to know the intricate workings.

3. Inheritance:

Definition: Mechanism by which one class (child/subclass) inherits the attributes and methods of another class(parent/ superclass)
Benefits: Promotes code reuse and establishes a natural hierarchy. It allows new classes to be defined based on existing classes.

4. Polymorphism


# What is a class?

A class in programming, particularly in oops, is a blueprint or template for creating objects. It defines a set of attributes and methods that the created objects will have. Classes are a fundamental concept in oop and are used to encapsulate data and fucntions that operate on that data
It can contain multiple functions +,-, divide, multiply
Key characteristics of a class:

1. Attributes: these are variables that hold the data or state of the class and its objects. Attributes are sometimes rreferred to as fields or properties.
2. Methods: These are functions defined within a class that describe within a class that describe the behaviors or actions that an object created from the class can perform.
3. Constructor: A special method called __init__ in python, used to initialize the attributes of the class when an object is created.
4. Encapsulation: Classes encapsulate data and methods, bundling them together. This helps in hiding the internal state of the object from the outside world and only exposing a controlled interface.



```python
class bike:
    def __init__(name,model,color,year):    # __init__ it is our constructor
        name.model = model
        name.color = color  # all three are our objects
        name.year = year

    def result(name):
        return f" your bike model {name.model} and color {name.color} {name.year}"   #this is our method

txt = bike("Tvs","black",2025)   # these are our attributes
 
print(txt.result())
```


# Inheritance

Definition: hiding the complex implementationdetails and showing only the necessary features of the object.


Benefits: simplifies the interface to interact with the object making it easier to use and understand without needing to know the intricate workings

```python
class vehicle:
    def __init__(self,make,model):
        self.make = make
        self.model = model

    def info(self):      #ye self isliye hai kyoki attribute k andar data store hota hai isliye self k andar sara data hai
        return f"{self.make} lives at {self.model}"

class car(vehicle):
    def __init__(self,make, model, year):
        super().__init__(make, model)     # super is used because it indicates ki is vali class mein kya lia hai humne upar vali parent class se
        self.year = year

    def car_info(self):
        return f"{self.year} {self.info()}"

my_car = car("Toyota", "Corolla", 2024)
print(my_car.car_info())
```

```python
class owner:
    def __init__(self,name,address):
        self.name = name
        self.address = address

    def info(self):      #ye self isliye hai kyoki attribute k andar data store hota hai isliye self k andar sara data hai
        return f"{self.name} lives at {self.address}"

class car(owner):
    def __init__(self,name, address,car_name,model):
        super().__init__(name,address)     # super is used because it indicates ki is vali class mein kya lia hai humne upar vali parent class se
        self.car_name = car_name
        self.model = model

    def car_info(self):
        return f"{self.info()} have {self.car_name} {self.model}"

my_car = car("DAKSH", "District road", "Mustang", 2024)
print(my_car.car_info())
```

# Abstraction

It is the concept of hiding the complex implementation details and showing only the necessary features.

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod      #@abstractmethod is a decorator indicating that the method 
                                     # must be implemented by any subclass of shape
                            #ye abstractmethod sirf clss mein he likh sakte hai or kahi randomly nahi
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

rect = Rectangle(10,20)
print(rect.area())
print(rect.perimeter())
```

```python
import re

pattern = re.compile(r'\d{3}-\d{2}-\d{4}')

#searching for the pattern

text = "My ssn is 123-45-6789."
match = re.search(pattern,text)

if match:
    print(f"found: {match.group()}")

#findign all matches

text = "contact number: 123-45-6789, 987-65-4321."

matches = re.findall(r'\d{3}-\d{2}-\d{4}', text)
print(matches)

#replacing patterns

next_text = re.sub(r'\d{3}-\d{2}-\d{4}', 'xxx-xx-xxxx', text)
print(next_text)
```
