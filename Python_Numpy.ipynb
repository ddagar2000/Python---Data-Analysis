# Subplot

```python
x = [1,2,3,4,5]
y1 = [2,3,5,7,11]
y2 = [1,4,6,8,10]

fig, (ax1, ax2) =  plt.subplots(2,1, figsize = (10,6))

#first subplot
ax1.plot(x,y1,'r-o')  # here r is color, - is the patten of graph i.e. line , o is the marker point in the graph
ax1.set_title('First Subplot')
ax1.set_xlabel('X-axis')
ax1.set_ylabel('Y-axis')

#second subplot
ax2.plot(x,y2,'g--o')  # here g is color, -- is the patten of graph i.e. line , ^ is the marker point in the graph
ax2.set_title('Second Subplot')
ax2.set_xlabel('X-axis')
ax2.set_ylabel('Y-axis')

plt.tight_layout()   #this is used to show gap between the subplots

plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np

x = np.linspace(-5, 5, 100)    #linspace divides the space into given parts, in this case it will divide 100 in 100 parts
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))   #we can also use cos or tan, but sin gives the most curves

fig = plt.figure()
ax = fig.add_subplot(111, projection ='3d')    #111 is the size of plot, it should be either 222 not 2 or 22
ax.plot_surface(x,y,z, cmap = 'viridis')

ax.set_title('3D Surface Plot')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')

plt.show()
```

# Stacked Bar Plot

```python
categories = ['A', 'B', 'C']
values1 = [3,7,2]
values2 = [5,3,8]

bar_width = 0.5
index = range(len(categories))

plt.bar(index, values1, bar_width, label = 'Group1', color = 'b')
plt.bar(index, values2, bar_width, bottom = values1, label = 'Group2', color = 'r')

plt.xticks(index, categories)
plt.title('Stacked Bar Plot')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.legend()
plt.show()
```

# Merging and Joining

```python
# Creating Data Frames to merge
df1 = pd.DataFrame({'key': ['A','B','C'], 'value': [1,2,3]})
df2 = pd.DataFrame({'key': ['B','C','D'], 'value': [4,5,6]})

#Merging dataframes
print(pd.merge(df1,df2,on='key'))     #inner join
print(pd.merge(df1,df2,on='key', how = 'left'))    #left join
print(pd.merge(df1,df2,on='key', how = 'outer'))    #outer join

#concatenating  DataFrames
print(pd.concat([df1, df2]))
```

# Grouping and Aggregating

```python
# CREATING A DataFrame to group
df_group = pd.DataFrame({
    'key' : ['A', 'B', 'A', 'B'],
    'data' : range(4)
})

#Grouping and applying aggregate functions
grouped = df_group.groupby('key')
print(grouped.sum())
print(grouped.mean())
print(grouped.size())
```
