```python
## Feature engineering
df['Date'] = df['Date_of_Journey'].str.split('/').str[0]
df['Month'] = df['Date_of_Journey'].str.split('/').str[1]
df['Year'] = df['Date_of_Journey'].str.split('/').str[2]
```

```python
df.info()
```

```python
df['Date'] = df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
```

```python
df.drop('Date_of_Journey', axis = 1, inplace = True)   # axis = 1 is our column and inplace = True is deleting permanent
```

```python
df.info()
```

```python
df['Arrival_Time'] = df['Arrival_Time'].apply(lambda x:x.split(' ')[0])   #ek function bana dia lambda jo space k dono taraf ki chize alag 
#kar raha hai, or jo [0] hai iska matlab space k left side ka 0th index choose kia hai
```

```python
df['Arrival_hour'] = df['Arrival_Time'].str.split(':').str[0]
df['Arrival_min'] = df['Arrival_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Arrival_hour'] = df['Arrival_hour'].astype(int)
df['Arrival_min'] = df['Arrival_min'].astype(int)
```

```python
df.info()
```

```python
df.drop('Arrival_Time', axis = 1, inplace = True)
```

```python
df.head(2)
```

```python
df['Dep_Time'] = df['Dep_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Dep_hour'] = df['Dep_Time'].str.split(':').str[0]
df['Dep_min'] = df['Dep_Time'].str.split(':').str[1]
```

```python
df.head(2)
```

```python
df['Dep_hour'] = df['Dep_hour'].astype(int)
df['Dep_min'] = df['Dep_min'].astype(int)
```
