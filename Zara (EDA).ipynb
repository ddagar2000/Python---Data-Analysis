```python
# outlier detection 
# for price

q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)

iqr = q3-q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr


outliers_price = df[(df['price']<lower) | (df['price']>upper)]    #  | is used for condition
outliers_price
```

```python
# outlier detection 
# for price

q1 = df['price'].quantile(0.25)
q3 = df['price'].quantile(0.75)

iqr = q3-q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr


outliers_price = df[(df['price']<lower) | (df['price']>upper)]    #  | is used for condition works as or condition
outliers_price.shape
```

```python
# for sales volume

# outlier detection 
# for price

q1 = df['sales_volume'].quantile(0.25)
q3 = df['sales_volume'].quantile(0.75)

iqr = q3-q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr


outliers_price = df[(df['sales_volume']<lower) | (df['sales_volume']>upper)]    #  | is used for condition
outliers_price.shape
```
